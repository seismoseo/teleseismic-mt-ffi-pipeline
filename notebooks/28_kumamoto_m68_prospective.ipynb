## 2. Moment tensor (MTUQ) — a strike-slip lesson: azimuth balance vs four-lobe aliasing
The independent MTUQ mechanism was **unstable on the first (25-station) subset** — and the failure mode is instructive. Strike-slip radiation repeats every 90° of azimuth; with 14 of 25 stations crowded into one 45° southern band, the long-period surface waves locked onto a mechanism rotated ~90° from the W-phase (Kagan 67–78°). Doubling to a **60-station azimuth-balanced set (gap 97°→80°) snapped the joint DC back to 130/88/7, Mw 6.80, Kagan 27.3°** — confirming the aliasing diagnosis. Body P+SH stays aliased (91°): teleseismic SH is unusable at this geometry.

| variant | stations / gap | mechanism | Mw | Kagan vs W-phase |
|---|---|---|---|---|
| DC, body P only | 25 / 97° | 94/59/−29 | (grid edge) | 27.1° |
| DC, body P+SH | 25 / 97° | 58/50/−83 | 6.30 | 67.1° |
| DC, joint | 25 / 97° | 184/82/11 | 6.65 | ~78° |
| full MT (surface) | 25 / 97° | 198/81/9 (γ=δ=0) | 6.60 | 78.0° |
| **DC, joint, 60 stn** | **60 / 80°** | **130/88/7** | **6.80** | **27.3°** |
| DC, body P+SH, 60 stn | 60 / 80° | 212/85/−2 | 6.75 | 91.1° |

Two pipeline rules frozen with this prediction: (1) for **strike-slip**, azimuthal *balance* matters more than station count; (2) the shallow-event "body-only" rule from Morocco/Mexico is a **dip-slip** cure (Kanamori–Given Mrt/Mrp indeterminacy) and does **not** transfer to vertical strike-slip, where long-period surface waves are well behaved given adequate azimuth sampling. Residual 27° precision is honestly below the 5–19° achieved for the thrust events — the mechanism statement for the frozen prediction therefore leans on the W-phase planes plus the relative WISP fit.

## 3. Finite fault (WISP) — both nodal planes, Mww-seeded, with the time-shift step
`auto_model` on both W-phase planes, seeded from the W-phase moment; the cross-correlation `shift_match` step improved the misfit 9% and 6% (standing rule: always shift).

| plane | strike/dip/rake | Mw | misfit (pre→post shift) | peak slip | T5-95 / Vr |
|---|---|---|---|---|---|
| **NP2 = Futagawa trend (preferred)** | **212/75/−164** | 6.75 | 0.1607 → **0.1507** | 1.88 m | 13.6 s / 2.05 |
| NP1 = conjugate | 118/74/−16 | 6.75 | 0.1667 → 0.1523 | 1.63 m | 14.8 s / 2.09 |

The tectonically expected right-lateral plane fits marginally (~1%) better — a weak discriminant, as in Morocco/Chile/Mexico, but at least pointing the same way as the tectonics.

## 4. Prospective prediction & caveats (compare with USGS FFM when posted)
**Prediction (frozen):**
- **Fault plane:** right-lateral **212/75/−164** (Futagawa–Hinagu trend) — weakly preferred by the fit, strongly by tectonic context.
- **Mw 6.75** (WISP, both planes) vs W-phase 6.8.
- **Slip:** a **single compact asperity**, peak ~1.9 m, ~2–16 km depth, centred at / slightly updip of the hypocentre; **no shallow surface-slip maximum**. A smaller sibling of the 2016 Kumamoto rupture.
- **Duration:** T5-95 ≈ 14 s, moment-rate peak at ~3.5 s; **Vr ≈ 2.1 km/s**.

**Caveats:** teleseismic-only; plane preference ~1% in misfit (weak); independent point-source precision limited to ~27° by residual azimuth geometry (documented above). USGS will likely add regional/InSAR data; per the Al Haouz study their model may be more compact than ours.

**Scorecard when USGS posts:** plane choice · Mw · peak slip amplitude and depth extent · single-vs-multiple asperities · duration / moment-rate shape · rupture velocity.
