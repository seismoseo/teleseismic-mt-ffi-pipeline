# 2026 Mw 6.9 North Sumatra (Pematangsiantar) — a 172-km-deep intraslab earthquake
Event `us6000tkyk`: 2026-08-15T10:54:51.8Z, 3.086°N 99.015°E, hypocentre 172.5 km, **Mww 6.9** (M0 2.62×10¹⁹ N·m, centroid 180.5 km, **98.6% DC**). Intraslab rupture in the subducted Indo-Australian (Sunda) slab beneath the Toba caldera region — *not* the megathrust, which is offshore and shallow here. W-phase planes: **NP1 208/58/1** (left-lateral, moderately dipping) and NP2 117/89/148 (near-vertical, right-lateral oblique).

No USGS finite-fault product exists for this event. Deep-event protocol from nbs 24/25/30: body band raised to 25–60 s, depth read from the body-only curve, and — since 180 km lies far below WISP's ~126.5 km surface-wave GF cap — **surface waves are dropped on both planes, making the two-plane comparison body-only and fair by construction**.

**Headline: this is the pipeline's cleanest event.** Misfits 0.080 (best in project), M0 recovered to <1%, and MTUQ independently recovers the W-phase mechanism to **Kagan 7.4°** — also the best in project.

## 4. Finite fault (WISP) — and why the fault had to be enlarged
The first-pass `auto_model` fits were superb (misfit 0.080–0.082) **but violated the edge-containment rule**: 29% (NP1) and 47% (NP2) of peak slip sat in the deepest fault row — the 69 × 57 km domain was too small down-dip. Per the nb25 rule, the fault was enlarged **identically on both planes** (→ 97 × 114 km) and both re-inverted and re-shifted:

| | first pass (69×57 km) | enlarged (97×114 km) |
|---|---|---|
| edge slip top/bottom | 2–4% / **29–47%** ✗ | 2–3% / **2–4%** ✓ |
| misfit NP1 / NP2 | 0.0816 / 0.0803 | 0.0829 / **0.0795** |
| plane margin | 1.6% (meaningless) | **4.1%, NP2 preferred** |
| peak slip | 0.53–0.57 m | 0.44–0.50 m |
| Δσ (effective-area) | 1.2 MPa | 0.65–0.83 MPa |

**The enlargement changed the conclusion**: with the boundary artifact removed the plane margin grew from a meaningless tie to a suggestive 4.1% favouring **NP2 (208/58/1)** — the moderately-dipping left-lateral plane, which is also the plane MTUQ independently converged on (strike 202°, dip 49–56°). Still short of the ~10% decisive threshold (SPdA), so: *suggestive, corroborated, not conclusive*.

Both planes recover **M0 = 2.64×10¹⁹ N·m vs W-phase 2.62×10¹⁹ (<1%)**, T5-95 ≈ 15.5 s, STF peak 6–7 s, and the STF is plane-independent (**r = 0.972**) — the signature of well-determined moment release on a weakly-determined plane.

## 5b. Is the double-peaked STF real? Is the rupture two asperities?
Two separate questions, with two different answers.

**The double peak is robust.** The second STF pulse appears in **4 of 5** independently derived models (both nodal planes × both fault sizes), and its prominence *grows* as the parameterization improves — 0.07 and 0.16 of peak in the original (edge-violating) faults, rising to 0.24 and **0.38** after the enlargement. Only the unshifted NP1 model shows a single peak. A feature that survives a plane swap, a domain change and a time-shift realignment, becoming clearer each time, is data-driven rather than a regularization artifact.

| model | peak 1 | peak 2 (prominence) |
|---|---|---|
| **NP2_big_shift (preferred)** | 6.3 s | 16.1 s (**0.38**) |
| NP1_big_shift | 7.3 s | 14.3 s (0.24) |
| NP2_shift (small fault) | 6.3 s | 16.0 s (0.16) |
| NP1_shift (small fault) | 6.9 s | 14.0 s (0.07) |
| NP1 (unshifted) | 6.7 s | — |

**Two comparable asperities: not supported.** Splitting the moment by rupture time, pulse 1 (trup < 11 s) carries **67%** of the moment at 165–194 km depth and pulse 2 (trup ≥ 11 s) carries **33%** at 191–207 km; their centroids are **20 km apart** (15 km horizontal, 13 km deeper) with a 9.3 s separation → apparent migration **2.15 km/s**, a legitimate rupture velocity (contrast the kinematically impossible 0.8 km/s of the Colombia M7.4, nb31). So the *kinematics* of a down-dip second pulse are coherent. But as a *spatially distinct asperity* it is marginal: at a 50%-of-peak threshold the second patch is a single subfault, and at 60% it vanishes. The defensible description is **a dominant asperity plus a weaker, deeper secondary pulse ~20 km down-dip** — not two comparable asperities.


```python
import numpy as np, json, os
E = os.path.join(B, 'event_2026_sumatra/20260815105451/ffm.0/NP2_big_shift')
txt = open(os.path.join(E, 'Solution.txt')).read().splitlines()
i0 = next(i for i, l in enumerate(txt) if 'slip' in l and 'rake' in l) + 1
A = np.array([[float(x) for x in l.split()[:11]] for l in txt[i0:] if len(l.split()) >= 11])
lat, lon, dep, slip, trup, mo = A[:,0], A[:,1], A[:,2], A[:,3]/100., A[:,7], A[:,10]
elat, elon = 3.0856, 99.0150
ce = (lon-elon)*111.32*np.cos(np.radians(lat)); cn = (lat-elat)*111.2
p1, p2 = trup < 11, trup >= 11
for lab, m in (('pulse 1 (trup<11 s)', p1), ('pulse 2 (trup>=11 s)', p2)):
    w = mo[m]/mo[m].sum()
    print(f'{lab}: {mo[m].sum()/mo.sum():5.0%} of moment | centroid depth {np.sum(w*dep[m]):3.0f} km, '
          f'{np.hypot(np.sum(w*ce[m]), np.sum(w*cn[m])):4.1f} km from epicentre')
w1, w2 = mo[p1]/mo[p1].sum(), mo[p2]/mo[p2].sum()
sep = np.hypot(np.sum(w2*ce[p2])-np.sum(w1*ce[p1]), np.sum(w2*cn[p2])-np.sum(w1*cn[p1]))
dsep = np.sum(w2*dep[p2]) - np.sum(w1*dep[p1])
dt = np.average(trup[p2], weights=mo[p2]) - np.average(trup[p1], weights=mo[p1])
print(f'separation {np.sqrt(sep**2+dsep**2):.0f} km ({sep:.0f} horizontal, {dsep:+.0f} depth) in {dt:.1f} s '
      f'-> apparent migration {np.sqrt(sep**2+dsep**2)/dt:.2f} km/s')
```

## 5c. Stress drop: convention matters by an order of magnitude

| convention | area | radius | Δσ |
|---|---|---|---|
| effective, (Σu)²/Σu² | 1827 km² | 24.1 km | **0.82 MPa** |
| slip > 15% of peak | 1479 km² | 21.7 km | 1.13 MPa |
| half-peak (> 50%) | 440 km² | 11.8 km | 6.98 MPa |

The spread is a direct consequence of smooth slip (mean 0.20 m over the effective area versus a 0.50 m peak). Quoting a single number without its convention would be misleading. A rigorous slip-weighted average of the on-fault static stress-change field (Ripperger & Mai 2007 kernel with Noda et al. 2013 energy-based averaging), computed separately, gives **≈0.9 MPa** for the preferred model — i.e. the *effective-area* convention tracks the rigorous value here, while the half-peak proxy overestimates it by nearly an order of magnitude. All of these are **lower bounds**: inversion smoothing suppresses short-wavelength slip and therefore stress.

**Rupture speed** is stable across models and interior to its allowed window (1.2–3.8 km/s, reference 3.0), so it is data-selected rather than a bound artifact:

| model | median | 16–84% | slip-weighted slope |
|---|---|---|---|
| **NP2_big_shift (preferred)** | **2.52 km/s** | 2.24–2.95 | 2.75 |
| NP1_big_shift | 2.60 | 2.35–2.87 | 2.68 |
| NP2_shift (small fault) | 2.50 | 2.06–2.72 | 2.48 |
| NP1_shift (small fault) | 2.38 | 2.14–2.66 | 2.36 |

With Vs ≈ 4.6 km/s at 180 km, **Vr/Vs ≈ 0.55** for the preferred model — the same slow, dissipative sub-shear regime found for the 2026 Colombia M7.4 (0.56).


**Depth (left panel) — an instructive split verdict.** The body-wave misfit curve is **flat to 0.3%** across 140–220 km, so depth is *not* resolved by the misfit minimum (its nominal minimum at 155 km is meaningless at that flatness). But **mechanism stability resolves it**: the recovered mechanism is chaotic at 140–170 km (Kagan 85–99°) and locks onto the W-phase from 180.5 km downward (5.5–18°). The body waves only fit coherently once the depth phases are timed correctly — which happens at the true centroid depth. This independently corroborates the USGS 180.5 km centroid over the 172.5 km hypocentre, and is reinforced by the WISP slip centroid landing at **181 km**.

**Rupture kinematics (preferred plane).** Slip centroid 8.4 km from the epicentre at azimuth 349° and 181 km depth; early→late moment centroid migrates 13.7 km toward **azimuth 354° (N)** while *deepening* 176→188 km, with 66%/23% forward/backward moment — a moderately unilateral northward, down-dip rupture. Realized front speed **median 2.52 km/s** (16–84%: 2.24–2.95), i.e. **Vr/Vs ≈ 0.55** at Vs ≈ 4.6 km/s — the same dissipative, sub-shear regime found for the Colombia M7.4 (0.56) and matching Poli et al.'s (2016) slow first-stage rupture rather than their fast runaway stage. Stress drop 0.65–0.83 MPa is low for an intraslab event.
