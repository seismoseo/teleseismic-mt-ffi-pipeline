## 2. The discovery that drives everything: a ~35-s emergent nucleation phase
The observed P wavetrains show a weak, emergent onset and the main impulsive packet only **~45–55 s after origin** (with pP/sP echoes at ~75–90 s). Independently, the **W-phase centroid time is 12:35:02.6 — 34.5 s after origin — and 29 km south** of the epicentre. The apparent nucleation-to-asperity velocity is ~0.8 km/s, far below any allowed rupture speed, so an origin-anchored finite fault is *kinematically impossible*: the first inversion could only place late moment at the fault edge (86% of peak at the domain boundary — a violated edge rule), and origin-cut MTUQ windows missed the main energy (Mw pinned at the grid edge).

**Fix (the operational Yeck-style relocation): re-anchor both analyses at the W-phase centroid in space *and* time.** The precursor becomes pre-window noise; the main rupture is imaged about its true origin. The invalid origin-anchored model is kept below as the diagnostic.

The body-wave depth curve bottoms at 70–90 km — shallower than the W-phase centroid — and the mechanism is stable nowhere except ~90 km, where it snaps to 225/71/4 (28.6°). ## 6b. Directivity — two-stage, unilateral southward + up-dip
(Correction note: the first push described the asperity as 'NE along strike' — an axis-misread of the WISP plot; the geographic computation below is authoritative and was corrected within the same pre-USGS freeze window.)

1. **Model-independent stage 1**: epicentre → W-phase centroid = **30 km toward azimuth 163° (SSE)**, 110 → 120.5 km, in ~34.5 s — apparent 0.9 km/s (the emergent nucleation phase).
2. **Preferred plane stage 2**: early→late moment centroid moves **27 km toward azimuth 220° (SW)** while shallowing 110 → 101 km; **75% of moment forward** vs 24% backward — predominantly unilateral SW + up-dip.
3. **Robustness**: up-dip shallowing appears on *both* planes; the horizontal direction on the conjugate is northward — inconsistent with the model-independent stage-1 offset, further disfavoring that (already rejected) plane.


```python
import numpy as np
elat, elon = 4.8436, -76.2422; clat, clon = 4.5854, -76.1613
def km(lon, lat, olon, olat):
    return (lon-olon)*111.32*np.cos(np.radians(lat)), (lat-olat)*111.2
for P, label in (('NP1_shift', 'preferred 226/72/33'), ('NP2_shift', 'conjugate 124/59/159')):
    txt = open(os.path.join(B, f'event_2026_colombia/20260810123502/ffm.0/{P}/Solution.txt')).read().splitlines()
    i0 = next(i for i, l in enumerate(txt) if 'slip' in l and 'rake' in l) + 1
    A = np.array([[float(x) for x in l.split()[:11]] for l in txt[i0:] if len(l.split()) >= 11])
    lat, lon, dep, trup, mo = A[:,0], A[:,1], A[:,2], A[:,7], A[:,10]
    w = mo/mo.sum(); ce, cn = km(lon, lat, clon, clat)
    az = np.degrees(np.arctan2(np.sum(w*ce), np.sum(w*cn))) % 360
    tmed = np.average(trup, weights=mo); e, l = trup <= tmed, trup > tmed
    dE = np.average(ce[l], weights=mo[l]) - np.average(ce[e], weights=mo[e])
    dN = np.average(cn[l], weights=mo[l]) - np.average(cn[e], weights=mo[e])
    azt = np.degrees(np.arctan2(dE, dN)) % 360
    u = np.sin(np.radians(azt))*ce + np.cos(np.radians(azt))*cn
    print(f'{label}: slip centroid {np.hypot(np.sum(w*ce),np.sum(w*cn)):.0f} km @ az {az:.0f} | '
          f'early->late {np.hypot(dE,dN):.0f} km @ az {azt:.0f}, depth '
          f'{np.average(dep[e],weights=mo[e]):.0f}->{np.average(dep[l],weights=mo[l]):.0f} km | '
          f'fwd/back {mo[u>2].sum()/mo.sum():.0%}/{mo[u<-2].sum()/mo.sum():.0%}')
dE, dN = km(clon, clat, elon, elat)
print(f'stage-1: {np.hypot(dE,dN):.0f} km @ az {np.degrees(np.arctan2(dE,dN))%360:.0f}, 110->120.5 km, ~34.5 s (apparent {np.hypot(dE,dN)/34.5:.1f} km/s)')
```

## 6c. Tectonic context: background seismicity, the Caldas tear, and the CARMA tomography
Background: 1,981 ComCat events (M≥4, depth ≥60 km, 1980–2026) and the CARMA finite-frequency dVp model of Cornthwaite et al. (2021, G-cubed; IRIS EMC `CARMA_nwSA_dVp_Tomo`) — the same array behind Sun et al. (2022, EPSL). Note the EMC file labels its units 'percent' but stores fractions; values here are rescaled (max |dVp| ≈ 9.9%, matching the papers).

**What the figure shows about our rupture:**
1. **The rupture is the Cauca cluster breaking.** Our >15%-of-peak slip zone (red) lies *inside* the tight NNE-trending band of intermediate-depth seismicity at 4–5.2°N, and in cross-section the slip patch (85–125 km) sits exactly on the east-dipping seismicity band that defines the local Nazca slab. The hypocentre sits at the patch's northern/updip edge; rupture propagated SSW **into the cluster interior, away from the tear**.
2. **The Caldas tear bounds the system.** The Cauca band terminates at ~5.15°N; between ~5.2° and 6.5°N there is a marked gap in deep seismicity separating it from the Bucaramanga domain — the segment (per Sun et al. 2022, plate) boundary. Our event ruptured to within a few tens of km of the tear on its southern side, consistent with the along-strike-compression (map-view strike-slip) mechanism.
3. **Bucaramanga is a different system.** The nest is the dense knot 407 km NE, underlain by the strong fast (blue) anomaly at 120 km — the slab Sun et al. (2022) identify as *Caribbean*, bending in two planes. Different plate, different stress concentrator.
4. **Tomography coverage, honestly:** the CARMA model's domain (2.5–15°N) nominally contains our event, but the imaged amplitude there is ~0.01% — the SW resolution edge (the array sits NE). At this location the *seismicity*, not the tomography, defines the slab; the well-resolved tomographic structure (fast Caribbean slab, segmentation) lies north-east.

## 6d. On-fault view: background seismicity projected onto the rupture plane, and prior M6+ mechanisms
Background events (M≥4, 1980–2026) within 25 km of the preferred plane (226/72), rotated into along-strike / down-dip coordinates over our slip model; prior M6+ cluster events with available moment tensors shown as beachballs with Kagan angles vs our W-phase mechanism.

**Findings:**
1. **The 2026 main rupture nucleated where the cluster's most recent M6 broke:** the 2019-03-23 M6.1 Versalles hypocentre projects within ~2 km of the plane, ~7 km from our centroid nucleation — but its mechanism (237/8/−40, Kagan 80°) is entirely different: the same volume fails on differently oriented structures, exactly the mechanism diversity Chang et al. (2019, JGR) documented for the cluster at M<5.
2. **Only the deep 1997 Roncesvalles pair (M6.8 at 199 km, M6.4 at 178 km) resemble our mechanism** (Kagan 35–36°) — oblique-reverse like ours, but 60–80 km deeper and ~40 km down-dip; the 1995 M6.6 Neira (120 km, in-cluster) was normal-type (Kagan 69°).
3. **The rupture plane cuts across the slab:** Chang et al.'s relocated slab seismic zone dips 33–43°; our plane dips 72°. The M7.4 broke a steep structure *within* the slab (as at San Pedro de Atacama), not the slab plane — consistent with reactivation of an inherited fault, and with the scattered (not plane-hugging) projection of background seismicity.
4. No prior M≥4 concentration outlines our main asperity — the patch that produced 1.6 m of slip was comparatively quiet at M≥4 (ComCat completeness caveat: SGC locations would sharpen this).
