## 3. Finite fault (WISP) — both nodal planes, Mww-seeded, with the time-shift step
`auto_model` on both W-phase planes, seeded from the W-phase moment (the key Al Haouz lesson). We
then applied the **cross-correlation time-shift step** (`shift_match`) — which `auto_model` does
**not** run (it's a separate manual step) — and re-inverted. This cut the misfit **13–16%** on both
planes with **Mw unchanged** (as on Al Haouz). Values below are the shifted models.

| plane | strike/dip | Mw | misfit (pre→post shift) | L×W / peak slip | slip depth / T5–95 |
|---|---|---|---|---|---|
| NP1 steep | 123/64 | 7.29 | 0.1120 → **0.0940** | 82×41 km / 3.43 m | 1–34 km / 19.2 s |
| **NP2 shallow (interface)** | **286/27** | **7.29** | 0.1114 → **0.0969** | 82×68 km / 2.54 m | 3–24 km / 20.3 s |

Both fit almost identically — teleseismic waveforms are a weak plane discriminant (as for Al Haouz
and Chile). For a subduction thrust the shallow NP2 is the megathrust interface and fits marginally
better; **we prefer NP2** but show both. The time-shift step sharpened the slip (NP2 peak 2.29→2.54 m).

## 5. Comparison with the USGS finite fault (posted *after* our freeze)
USGS (D. Goldberg) posted their finite fault on 2026-07-17 — and it is **also teleseismic-only**
(BODY 63, SURF 52, InSAR 0), a true like-for-like comparison.

| | USGS (Goldberg) | Ours (NP2_shift) | assessment |
|---|---|---|---|
| **Fault plane** | 286/27/75 | 286/27/75 | **exact match** ✓ |
| **Mw** | 7.29 | 7.29 | exact ✓ |
| Peak slip | 2.85 m | 2.54 m | close (−11%) |
| **Slip depth (>15%)** | 4–25 km | 3–24 km | **excellent match** ✓ |
| Moment-rate peak | 10.2 s | 10.2 s | exact ✓ |
| **Data used** | teleseismic-only | teleseismic-only | match ✓ |
| Fault template | 125×65 km | 82×68 km | USGS larger |
| Rupture velocity | 1.85 km/s | 2.08 km/s | ours faster |
| Duration T5–95 | 35 s | 20 s | USGS longer |

**Blind-test outcome — the fundamentals validated.** Same nodal plane (we correctly preferred the
shallow megathrust interface), same Mw, same slip depth, same peak-slip magnitude, same main-pulse
timing — both teleseismic-only. The differences are secondary modelling choices: USGS used a larger
fault template (125 km) + slower Vr → a longer, more spread-out rupture; ours is more compact.

```python
display(Image(os.path.join(B,'event_2026_mexico/fig_usgs_comparison.png')))
```

The main pulse (most of the moment) is nearly identical in timing and amplitude, and the slip
depth distributions nearly overlap (both peak ~16 km). USGS spreads the last ~15% of moment over a
longer tail (bigger fault + slower Vr).

**Where our forecast missed (honest).** Our caveats predicted the *opposite* of what USGS did: we
expected them to add InSAR and be *more compact*. Instead they are teleseismic-only and *longer*.
So our prediction was, if anything, conservatively compact — mechanism, moment, slip depth and
main-pulse timing all validated; total rupture length/duration is a genuine model-choice difference
(template size + Vr) on which teleseismic data are weakly constraining.

## 6. A pipeline fix prompted by this event — teleseismic station coverage
The USGS model used **63 P stations with near-full azimuthal coverage**; our first run used only
**15** (an 87° Pacific gap). That exposed a real bug in our fetcher (`wisp_fetch_fallback.py`): it
**required all three components**, silently dropping every BHZ-only (P) station, and used
per-station requests with a bare `except: continue` that dropped any timeout / not-yet-arrived data
with no record. The GSN inventory had **81 usable stations** at 30–90°; we kept 15.

The rewritten **`fetch_teleseismic.py`** (new pipeline standard) queries the GSN + a globally
distributed backbone, **keeps a station on BHZ alone**, uses a single bulk request with retry, and
**logs every drop + the azimuthal coverage**. Result on this event: the fetcher now provides
**54 stations (vs 15)** and the azimuthal gap shrinks **87° → 60°** (WISP's auto quality-control
then accepts 22 P of the 54; total data 2364 → 3216 values, +36%).

### Did fuller coverage bring us to USGS? An honest controlled test — no.

| model | fault | duration T5–95 | misfit |
|---|---|---|---|
| sparse (15 stn) | 82 km | 20.8 s | 0.1114 |
| **full coverage (54 stn)** | 82 km | 22.3 s | **0.1108** |
| full coverage **+ USGS's exact 125×65 km template** | 82 km *effective* | 23.9 s | 0.1104 |
| **USGS** | 125 km | 35.0 s | — |

Fuller coverage improved the fit slightly but **barely moved the rupture length**. Even forcing
USGS's *exact* 125-km template onto the full-coverage data, our inversion concentrates slip in
~82 km / ~24 s — the teleseismic data prefer a compact rupture even on USGS's big canvas. So the
under-sampling was a genuine weakness (now fixed) but **not** the cause of the length difference;
that is an inversion-parameterization choice on which teleseismic data are weakly constraining.
